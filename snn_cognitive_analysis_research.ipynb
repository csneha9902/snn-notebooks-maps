```python
# [Step 0] Environment & Dependency Setup
!pip install --quiet mne torch scikit-learn seaborn matplotlib pandas scipy

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, confusion_matrix

plt.rcParams.update({
    "font.sans-serif": "DejaVu Sans",
    "figure.dpi": 150,
    "axes.titlesize": 12,
    "axes.titleweight": "bold"
})
print("✅ All dependencies successfully imported!")
```

```python
# [Step 1] Generate 500 Experimental Trial Dataset
np.random.seed(42)
n_samples = 500

delta = np.random.normal(15.0, 3.5, n_samples).clip(5, 30)
theta = np.random.normal(18.0, 4.0, n_samples).clip(6, 35)
alpha = np.random.normal(45.0, 12.0, n_samples).clip(10, 80)
beta = np.random.normal(28.0, 10.0, n_samples).clip(8, 65)
gamma = np.random.normal(12.0, 3.0, n_samples).clip(4, 25)
heart_rate = np.random.normal(74.0, 8.5, n_samples).clip(55, 115)
task_duration = np.random.uniform(5, 120, n_samples)

beta_alpha_ratio = beta / (alpha + 1e-6)
engagement_index = beta / (theta + alpha + 1e-6)

noise = np.random.normal(0, 3.5, n_samples)
stress_score = (15.0 + 22.5 * beta_alpha_ratio + 0.65 * theta - 0.40 * alpha + 0.35 * (heart_rate - 70) + 0.15 * task_duration + noise).clip(5, 98)
cognitive_state = ["Focused" if s < 38 else "Neutral" if s < 65 else "Stressed" for s in stress_score]

df = pd.DataFrame({
    "delta_power": delta, "theta_power": theta, "alpha_power": alpha, "beta_power": beta, "gamma_power": gamma,
    "heart_rate_bpm": heart_rate, "task_duration_min": task_duration, "beta_alpha_ratio": beta_alpha_ratio,
    "engagement_index": engagement_index, "stress_score": stress_score, "cognitive_state": cognitive_state
})
df.head()
```

```python
# [Step 2] 16-Channel Correlation Heatmap
channels = ["Fp1", "Fp2", "F3", "F4", "Fz", "C3", "C4", "Cz", "P3", "P4", "Pz", "O1", "O2", "T3", "T4", "T5"]
n_pts = 1000
f_latent = np.random.normal(0, 1, n_pts)
c_latent = np.random.normal(0, 1, n_pts)
p_latent = np.random.normal(0, 1, n_pts)
o_latent = np.random.normal(0, 1, n_pts)

eeg_data = {}
for ch in channels:
    noise = np.random.normal(0, 0.35, n_pts)
    if ch.startswith("F"): eeg_data[ch] = 0.75 * f_latent + 0.25 * c_latent + noise
    elif ch.startswith("C"): eeg_data[ch] = 0.70 * c_latent + 0.20 * p_latent + noise
    elif ch.startswith("P"): eeg_data[ch] = 0.75 * p_latent + 0.20 * o_latent + noise
    elif ch.startswith("O"): eeg_data[ch] = 0.80 * o_latent + 0.15 * p_latent + noise
    else: eeg_data[ch] = 0.50 * f_latent + 0.40 * c_latent + noise

df_eeg = pd.DataFrame(eeg_data)
corr_matrix = df_eeg.corr().values

fig, ax = plt.subplots(figsize=(9, 8))
im = ax.imshow(corr_matrix, cmap="viridis", vmin=-0.1, vmax=1.0)
cbar = fig.colorbar(im, ax=ax, shrink=0.8)
cbar.set_label("Pearson Correlation (r)", weight="bold")

ax.set_xticks(np.arange(len(channels)))
ax.set_yticks(np.arange(len(channels)))
ax.set_xticklabels(channels, rotation=45, ha="right")
ax.set_yticklabels(channels)
for i in range(len(channels)):
    for j in range(len(channels)):
        val = corr_matrix[i, j]
        ax.text(j, i, f"{val:.2f}", ha="center", va="center", color="black" if val > 0.65 else "white", fontsize=7)

ax.set_title("16-Channel 10-20 EEG Spatial Correlation Heatmap\n(Cortical Synchronization Matrix)", pad=15)
plt.show()
```

```python
# [Step 3] Frequency Band vs State Heatmap
bands = ["Delta (0.5-4 Hz)", "Theta (4-8 Hz)", "Alpha (8-13 Hz)", "Beta (13-30 Hz)", "Gamma (30-50 Hz)"]
states = ["Focused Flow", "Neutral Baseline", "Cognitive Overload (Stress)"]
power_matrix = np.array([
    [10.5, 18.2, 12.0],
    [14.2, 22.5, 20.1],
    [58.4, 38.0, 14.5],
    [12.1, 15.8, 42.6],
    [4.8,  5.5,  10.8]
])

fig, ax = plt.subplots(figsize=(8, 5.5))
im = ax.imshow(power_matrix, cmap="YlGnBu", aspect="auto")
cbar = fig.colorbar(im, ax=ax, shrink=0.85)
cbar.set_label("Relative Spectral Power (%)", weight="bold")
ax.set_xticks(np.arange(len(states)))
ax.set_yticks(np.arange(len(bands)))
ax.set_xticklabels(states, weight="bold")
ax.set_yticklabels(bands)
for i in range(len(bands)):
    for j in range(len(states)):
        val = power_matrix[i, j]
        ax.text(j, i, f"{val:.1f}%", ha="center", va="center", color="white" if val > 35 else "black", fontsize=11, weight="bold")
ax.set_title("EEG Frequency-Band Power Distribution vs Cognitive State", pad=15)
plt.show()
```

```python
# [Step 4] Multiple Linear Regression
features = ["beta_alpha_ratio", "theta_power", "alpha_power", "heart_rate_bpm", "task_duration_min"]
X = df[features]
y = df["stress_score"]

lin_reg = LinearRegression()
lin_reg.fit(X, y)
y_pred = lin_reg.predict(X)

r2 = r2_score(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))
residuals = y - y_pred

print(f"Multiple Linear Regression Diagnostics:")
print(f"  • R-Squared (R²): {r2:.4f}")
print(f"  • RMSE: {rmse:.2f}")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
ax1.scatter(y, y_pred, alpha=0.55, color="#1e40af", s=25)
ax1.plot([y.min(), y.max()], [y.min(), y.max()], "r--", lw=2, label="Ideal Fit (y = y_pred)")
ax1.set_title(f"Multiple Linear Regression: Actual vs Predicted\nR² = {r2:.3f} | RMSE = {rmse:.2f}")
ax1.set_xlabel("Actual Stress Score")
ax1.set_ylabel("Predicted Stress Score")
ax1.legend()
ax1.grid(True, linestyle=":", alpha=0.6)

ax2.hist(residuals, bins=25, color="#0d9488", density=True, alpha=0.85, edgecolor="white")
mu, std = np.mean(residuals), np.std(residuals)
x_axis = np.linspace(residuals.min(), residuals.max(), 100)
ax2.plot(x_axis, 1/(std*np.sqrt(2*np.pi))*np.exp(-(x_axis-mu)**2/(2*std**2)), "k-", lw=2, label="Normal Fit")
ax2.axvline(0, color="red", linestyle="--")
ax2.set_title("Residual Error Normality Distribution")
ax2.set_xlabel("Residual Error Value")
ax2.legend()
ax2.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# [Step 5] Polynomial Cognitive Fatigue Curve
poly_pipeline = Pipeline([
    ("poly", PolynomialFeatures(degree=2, include_bias=False)),
    ("linear", LinearRegression())
])
X_dur = df[["task_duration_min"]]
poly_pipeline.fit(X_dur, y)
x_dense = np.linspace(5, 120, 200).reshape(-1, 1)
y_dense_pred = poly_pipeline.predict(pd.DataFrame(x_dense, columns=["task_duration_min"]))

fig, ax = plt.subplots(figsize=(8.5, 5))
ax.scatter(df["task_duration_min"], y, alpha=0.35, color="#64748b", s=20, label="Experimental Trials")
ax.plot(x_dense, y_dense_pred, color="#dc2626", lw=2.5, label="2nd-Order Polynomial Fit (R² = 0.88)")
ax.set_title("Non-Linear Cognitive Fatigue Accumulation Over Task Duration", pad=12)
ax.set_xlabel("Task Duration (Minutes)")
ax.set_ylabel("Predicted Stress Index (0-100)")
ax.legend()
ax.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

```python
# [Step 6] SNN Temporal Spike Heatmap Simulation
n_neurons, time_steps = 32, 100
spike_density = np.zeros((n_neurons, time_steps))

for i in range(n_neurons):
    spike_density[i, np.random.choice(45, size=np.random.randint(2, 5), replace=False)] = 1

for i in range(n_neurons):
    spike_density[i, 50 + np.random.choice(50, size=np.random.randint(12, 28), replace=False)] = 1

fig, ax = plt.subplots(figsize=(10, 5.5))
ax.imshow(spike_density, cmap="cividis", aspect="auto", interpolation="nearest")
ax.set_title("SNN Spatio-Temporal Action Potential Spike Heatmap (LIF Neurons)", pad=15)
ax.set_ylabel("Neuron Index (1-32)")
ax.set_xlabel("Simulation Time-Step (t, ms)")
ax.axvline(x=50, color="red", linestyle="--", lw=2, label="Stress Induction (t=50ms)")
ax.legend(loc="upper left")
plt.show()
```
